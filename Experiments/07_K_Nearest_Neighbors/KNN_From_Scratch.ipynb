```python
import numpy as np
from collections import Counter
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
```

```python
X, Y = load_iris(return_X_y=True)
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
print(X_train.shape, X_test.shape)
```

```python
def knn_predict(X_train, Y_train, test_point, k=3):
    # 1. distance from test point to every training point
    distances = np.sqrt(((X_train - test_point) ** 2).sum(axis=1))

    # 2. indices of the k smallest distances
    k_idx = distances.argsort()[:k]

    # 3. majority vote among those k labels
    return Counter(Y_train[k_idx]).most_common(1)[0][0]
```

```python
# predict one point
print(knn_predict(X_train, Y_train, X_test[0], k=3), "actual:", Y_test[0])
```

```python
# predict all test points + accuracy
Y_pred = np.array([knn_predict(X_train, Y_train, p, k=3) for p in X_test])
print("accuracy:", (Y_pred == Y_test).mean())
```
